```python
## Popularity Based Recommender System
```

```python
ratings_name = Ratings.merge(Books, on='ISBN')
```

```python
ratings_name
```

```python
num_rating_df = ratings_name.groupby('Book-Title').count()['Book-Rating'].reset_index()
num_rating_df.rename(columns={'Book-Rating': 'num_ratings'}, inplace=True)
num_rating_df
```

```python
avg_rating_df = ratings_name.groupby('Book-Title')['Book-Rating'].mean().reset_index()
avg_rating_df.rename(columns={'Book-Rating': 'avg_rating'}, inplace=True)
avg_rating_df

```

```python
popular_df = num_rating_df.merge(avg_rating_df, on='Book-Title')
popular_df
```

```python
popular_df = popular_df[popular_df['num_ratings']>=250].sort_values('avg_rating',ascending = False).head(50) 
```

```python
popular_df.shape
```

```python
popular_df = popular_df.merge(Books, on='Book-Title').drop_duplicates('Book-Title')[['Book-Title','Book-Author','Image-URL-M','num_ratings','avg_rating']]
```

Collaborative Filtering Based Recommender System

```python
x = ratings_name.groupby('User-ID').count()['Book-Rating']>200
users_rates_more = x[x].index
users_rates_more
```

```python
filtered_rating = ratings_name[ratings_name['User-ID'].isin(users_rates_more)]
filtered_rating
```

```python
y = filtered_rating.groupby('Book-Title').count()['Book-Rating']>=50
famous_books = y[y].index
famous_books
```

```python
final_ratings = filtered_rating[filtered_rating['Book-Title'].isin(famous_books)]
```

```python
pt = final_ratings.pivot_table(index='Book-Title', columns='User-ID', values='Book-Rating')
```

```python
pt.fillna(0,inplace=True)          # users who rates 200 books and more, books which have more then 50 ratings or 50 
pt  
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity_scores = cosine_similarity(pt)
```

```python
similarity_scores
```

```python
def recommend(book_name) :
     # index fetch
     index = np.where(pt.index==book_name)[0][0]
     data = []
     similar_items = sorted(list(enumerate(similarity_scores[index])),key=lambda x:x[1], reverse=True)[1:6]
     for i in similar_items:

         item = []
         temp_df = Books[Books['Book-Title'] == pt.index[i[0]]]
         item.extend(temp_df.drop_duplicates('Book-Title')['Book-Title'].values)
         item.extend(temp_df.drop_duplicates('Book-Title')['Book-Author'].values)
         item.extend(temp_df.drop_duplicates('Book-Title')['Image-URL-M'].values)

         data.append(item)
     return data

        
```

```python
recommend('Zoya')
```
